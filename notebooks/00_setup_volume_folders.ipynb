```python
# Create the landing/processed folder structure in both raw-file volumes.
# Safe to re-run: mkdirs does nothing if a folder already exists.

volumes = ["/Volumes/health/bronze/fhir_raw", "/Volumes/health/bronze/hl7_raw"]

for vol in volumes:
    for folder in ["landing", "processed"]:
        dbutils.fs.mkdirs(f"{vol}/{folder}")

for vol in volumes:
    print(vol, [f.name for f in dbutils.fs.ls(vol)])
```

```python
fhir = dbutils.fs.ls("/Volumes/health/bronze/fhir_raw/landing/")
hl7  = dbutils.fs.ls("/Volumes/health/bronze/hl7_raw/landing/")
print("FHIR files:", len(fhir), " HL7 files:", len(hl7))
print("Odd names:", [f.name for f in fhir + hl7 if "[" in f.name or "(" in f.name])
```

```python
src = "/Volumes/health/bronze/fhir_raw/processed/"
dst = "/Volumes/health/bronze/fhir_raw/landing/"
files = dbutils.fs.ls(src)
for f in files:
    dbutils.fs.mv(f.path, dst + f.name)
print("Moved back", len(files), "files")   # expect 114
```
