# 00_config
Shared names and paths for every notebook. Other notebooks load it with `%run ./00_config`.

```python
CATALOG = "health"
BRONZE, SILVER, GOLD = "bronze", "silver", "gold"

FHIR_VOLUME = f"/Volumes/{CATALOG}/{BRONZE}/fhir_raw"
HL7_VOLUME  = f"/Volumes/{CATALOG}/{BRONZE}/hl7_raw"

def landing(volume):   return f"{volume}/landing/"
def processed(volume): return f"{volume}/processed/"

def move_to_processed(volume, file_names):
    """Move loaded files out of landing so the next run only sees new files."""
    for name in file_names:
        dbutils.fs.mv(landing(volume) + name, processed(volume) + name)
    print(f"Moved {len(file_names)} files to {processed(volume)}")

print("Config loaded:", CATALOG, FHIR_VOLUME, HL7_VOLUME)
```
